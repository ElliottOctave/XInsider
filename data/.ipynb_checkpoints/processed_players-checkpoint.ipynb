```python
import csv
import io

input_filename = "players.csv"
output_filename = "players_cleaned.csv"

cleaned_rows = []

with open(input_filename, mode="r", encoding="utf-8") as infile:
    for line in infile:
        line = line.strip()

        # If the whole line is quoted, clean it manually
        if line.startswith('"') and line.endswith('"'):
            line = line[1:-1]  # Remove outer quotes
            line = line.replace('""', '"')  # Fix escaped quotes

        # Use csv.reader to parse this cleaned line into fields
        reader = csv.reader(io.StringIO(line), delimiter=',', quotechar='"', skipinitialspace=True)
        parsed_row = next(reader)
        cleaned_rows.append(parsed_row)

# Write out cleaned data uniformly
with open(output_filename, mode="w", encoding="utf-8", newline='') as outfile:
    writer = csv.writer(outfile, quoting=csv.QUOTE_MINIMAL)
    writer.writerows(cleaned_rows)

print(f"✅ Uniform and cleaned dataset saved to: {output_filename}")
```
