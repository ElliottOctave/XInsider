```python
import pandas as pd

events = pd.read_csv("game_events.csv")
players = pd.read_csv("players.csv")
appearances = pd.read_csv("appearances.csv")

events['date'] = pd.to_datetime(events['date'], errors='coerce')
events['year'] = events['date'].dt.year
appearances['date'] = pd.to_datetime(appearances['date'], errors='coerce')
appearances['year'] = appearances['date'].dt.year

events['description'] = events['description'].fillna('')
is_own_goal = events['description'].str.lower().str.contains("own-goal")

goals = events[(events['type'] == 'Goals') & (~is_own_goal)]
yellow_cards = events[(events['type'] == 'Cards') & (events['description'].str.lower().str.contains('yellow'))]
red_cards = events[(events['type'] == 'Cards') & (events['description'].str.lower().str.contains('red'))]

assists = events[
    (events['type'] == 'Goals') &
    (events['player_assist_id'].notna()) &
    (events['player_assist_id'] != '')
].copy()

assists['player_assist_id'] = assists['player_assist_id'].astype(float).astype(int).astype(str)
assists['year'] = pd.to_datetime(assists['date'], errors='coerce').dt.year

for df in [goals, yellow_cards, red_cards, appearances]:
    df['player_id'] = df['player_id'].astype(str)
players['player_id'] = players['player_id'].astype(str)

goals_grouped = goals.groupby(['player_id', 'year']).size().rename("nr_of_goals")
yellow_grouped = yellow_cards.groupby(['player_id', 'year']).size().rename("yellow_cards")
red_grouped = red_cards.groupby(['player_id', 'year']).size().rename("red_cards")

assists_grouped = assists.groupby(['player_assist_id', 'year']).size().reset_index(name="assists")
assists_grouped = assists_grouped.rename(columns={'player_assist_id': 'player_id'})
assists_grouped['player_id'] = assists_grouped['player_id'].astype(str)
assists_grouped = assists_grouped.set_index(['player_id', 'year'])['assists']

minutes_grouped = appearances.groupby(['player_id', 'year'])['minutes_played'].sum().rename("minutes")

stats = pd.concat([
    goals_grouped,
    assists_grouped,
    yellow_grouped,
    red_grouped,
    minutes_grouped
], axis=1).fillna(0).reset_index()

for col in ['nr_of_goals', 'assists', 'yellow_cards', 'red_cards', 'minutes']:
    stats[col] = stats[col].astype(int)

stats = stats.merge(players[['player_id', 'name']], on='player_id', how='left')
stats = stats.rename(columns={"name": "player_name"})

stats.to_csv("player_stats.csv", index=False)
```
