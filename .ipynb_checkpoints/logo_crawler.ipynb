```python
import requests
from bs4 import BeautifulSoup
import csv
import time

input_csv = "clubs.csv"
output_csv = "club_logos.csv"

scraped_clubs = set()

with open(input_csv, newline='', encoding='utf-8') as infile, \
     open(output_csv, mode='w', newline='', encoding='utf-8') as outfile:

    reader = csv.DictReader(infile)
    fieldnames = ['club_id', 'club_name', 'logo_url']
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)
    writer.writeheader()

    for row in reader:
        club_id = row['club_id'].strip()
        club_name = row['name'].strip()
        url = row['url'].strip()

        if not url or club_id in scraped_clubs:
            continue

        try:
            headers = {"User-Agent": "Mozilla/5.0"}
            response = requests.get(url, headers=headers, timeout=10)

            if response.status_code == 200:
                soup = BeautifulSoup(response.text, 'html.parser')
                img_tag = soup.select_one('img[src*="/images/wappen/head/"]')

                if img_tag:
                    logo_url = img_tag['src']
                    writer.writerow({
                        'club_id': club_id,
                        'club_name': club_name,
                        'logo_url': logo_url
                    })
                    scraped_clubs.add(club_id)
                else:
                    print(f" Logo not found for {club_name}")
            else:
                print(f"Failed to fetch {url} (status {response.status_code})")

        except Exception as e:
            print(f"Error scraping {club_name}: {e}")

        time.sleep(1)
```
