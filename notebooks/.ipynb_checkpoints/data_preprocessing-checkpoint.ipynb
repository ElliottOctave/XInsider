# Data Preprocessing

This file contains the preprocessing for the players visualizations (player list and player details)

### Player summary

Get a summary of each player: goals, assists, club...

```python
import re
import pandas as pd
from io import StringIO

# Pre-clean CSV
with open("../data/players.csv", "r", encoding="utf-8") as f:raw_lines = f.readlines()

# Remove outer quotes if present and unescape inner quotes
cleaned_lines = []
for line in raw_lines:
    line = line.strip()

    # Remove leading/trailing quotes if the entire line is quoted
    if line.startswith('"') and line.endswith('"'):
        line = line[1:-1]

    # Replace inner double double-quotes with single quote (""South Korea"" to "South Korea")
    line = re.sub(r'""', '"', line)
    cleaned_lines.append(line)

# Load cleaned content into pandas
cleaned_csv = StringIO("\n".join(cleaned_lines))
players = pd.read_csv(cleaned_csv)
# South Korea is the only country with a comma in datasets so we change this
players['country_of_birth'] = players['country_of_birth'].replace('Korea, South', 'South Korea')
players['country_of_citizenship'] = players['country_of_citizenship'].replace('Korea, South', 'South Korea')

# Load other necessay datasets
game_events = pd.read_csv("../data/game_events.csv")
appearances = pd.read_csv("../data/appearances.csv")
club_logos = pd.read_csv("../data/club_logos.csv")
competition_logos = pd.read_csv("../data/competition_logos.csv")

# Preprocess game_events
# Extract goals, assist, yellow and red cards
goal_events = game_events[game_events['type'] == 'Goals']
goals = goal_events.groupby('player_id').size().rename('goals')
assists = goal_events['player_assist_id'].value_counts().rename('assists')
cards = game_events[game_events['type'] == 'Cards']
yellow_cards = cards[cards['description'].str.contains("Yellow", na=False)].groupby('player_id').size().rename('yellow_cards')
red_cards = cards[cards['description'].str.contains("Red", na=False)].groupby('player_id').size().rename('red_cards')

# Put all the info together
app_stats = appearances.groupby('player_id').agg({
    'goals': 'sum',
    'assists': 'sum',
    'red_cards': 'sum',
    'yellow_cards': 'sum',
    'minutes_played': 'sum',
}).fillna(0)

# Count total appearances and minutes played
unique_appearances = appearances.drop_duplicates(subset=['player_id', 'game_id'])
total_appearances = unique_appearances.groupby('player_id').size().rename('appearances')
total_minutes = total_appearances * 90
total_minutes.name = 'total_minutes'

# Combine the info together
event_stats = pd.concat([goals, assists, yellow_cards, red_cards], axis=1).fillna(0)
event_stats.index = event_stats.index.astype(int)
combined_stats = event_stats.add(app_stats, fill_value=0).fillna(0)
combined_stats = combined_stats.join(total_appearances)
combined_stats = combined_stats.join(total_minutes)
combined_stats = combined_stats.reset_index().rename(columns={'index': 'player_id'})

# Select the data we need and merge with necessary datasets
player_info = players[[ 
    'player_id', 'name', 'country_of_birth', 
    'country_of_citizenship', 'date_of_birth', 'position', 'foot', 
    'height_in_cm', 'current_club_id', 'current_club_name', 'current_club_domestic_competition_id', 'market_value_in_eur', 
    'highest_market_value_in_eur', 'image_url',
]]
player_info['player_id'] = pd.to_numeric(player_info['player_id'], errors='coerce')
player_info = player_info.dropna(subset=['player_id'])
player_info['player_id'] = player_info['player_id'].astype(int)

# Merge with club logos
club_logos['club_id'] = pd.to_numeric(club_logos['club_id'], errors='coerce')
club_logos = club_logos.dropna(subset=['club_id'])
club_logos['club_id'] = club_logos['club_id'].astype(int)
player_info = pd.merge(
    player_info,
    club_logos.rename(columns={'club_id': 'current_club_id', 'logo_url': 'club_logo_url'}),
    on='current_club_id',
    how='left'
)

# Merge with competition logos
player_info = pd.merge(
    player_info,
    competition_logos,
    left_on='current_club_domestic_competition_id',
    right_on='competition_id',
    how='left'
)

player_summary = pd.merge(player_info, combined_stats, on='player_id', how='left')
player_summary = player_summary.fillna(0)

# Convert date column in appearances to datetime
appearances['date'] = pd.to_datetime(appearances['date'], errors='coerce')

# Add first match year and last match year (for filtering)
match_years = appearances.groupby('player_id')['date'].agg(['min', 'max'])
match_years['first_match_year'] = match_years['min'].dt.year
match_years['last_match_year'] = match_years['max'].dt.year
match_years = match_years[['first_match_year', 'last_match_year']]

# Merge match years into player summary
player_summary = player_summary.merge(match_years, left_on='player_id', right_index=True, how='left')
transfers = pd.read_csv("../processed_data/transfers_preprocessed.csv")
players_with_transfers = transfers['player_id'].dropna().astype(int).unique()

# Filter players to only those with a transfer history
players = player_summary[player_summary['player_id'].isin(players_with_transfers)]
players = players.sort_values(by='highest_market_value_in_eur', ascending=False)

players.to_csv("../processed_data/player_summary.csv", index=False)

players.head(10)
```

### Count of players positions

```python
# Load the CSV file
df = pd.read_csv("data/game_lineups.csv")

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Filter only starting players (who actually played)
starting_players = df[df["type"] == "starting_lineup"]

# Count matches per player per position
position_counts = (
    starting_players
    .groupby(["player_id", "position"])
    .agg(matches_played=("game_id", "nunique"))
    .reset_index()
)

# Pivot: one row per player, one column per position
positions = position_counts.pivot(index="player_id", columns="position", values="matches_played").fillna(0).astype(int)
player_names = df[["player_id", "player_name"]].drop_duplicates()
positions = positions.reset_index().merge(player_names, on="player_id", how="left")

# Reorder columns: player_id, player_name, position columns
cols = ["player_id", "player_name"] + [col for col in positions.columns if col not in ["player_id", "player_name"]]
positions = pivoted[cols]

# Show the final table
positions.to_csv("../processed_data/position_count.csv", index=False)
```

### Winning club competition

For each competition, get the club that won it (and some other info)

```python
# Load the CSV files
games = pd.read_csv('../data/games.csv')
clubs = pd.read_csv('../data/clubs.csv')
competition_logos = pd.read_csv('../data/competition_logos.csv')

# Ensure that club_id is treated as an integer
clubs['club_id'] = clubs['club_id'].astype(int)

# Filter the games for finals
finals_games = games[games['round'].str.lower() == 'final']

# Create a function to determine the winner based on scores
def get_winner(game):
    if game['home_club_goals'] > game['away_club_goals']:
        return game['home_club_id']
    elif game['away_club_goals'] > game['home_club_goals']:
        return game['away_club_id']
    else:
        return None

# Identify the winner for each final game
finals_games['winner_id'] = finals_games.apply(get_winner, axis=1)
finals_games['winner_id'] = finals_games['winner_id'].astype('Int64')  # Using 'Int64' to handle NaN

# Merge with clubs to get the club name
finals_games_with_club_name = finals_games.merge(clubs[['club_id', 'name']], left_on='winner_id', right_on='club_id', how='left')

# Get the competition name and cup image from competition_logos
finals_games_with_competition = finals_games_with_club_name.merge(
    competition_logos[['competition_id', 'competition_name', 'cup_image_url']],
    left_on='competition_id',
    right_on='competition_id',
    how='left'
)

# Extract the year from the date column
finals_games_with_competition['year'] = pd.to_datetime(finals_games_with_competition['date']).dt.year

# Create the results dataframe with columns
final_results = finals_games_with_competition[['competition_name', 'name', 'club_id', 'year', 'cup_image_url']]
final_results = final_results.dropna(subset=['club_id'])
final_results['club_id'] = final_results['club_id'].astype(int)

# Save
final_results.to_csv("../processed_data/club_competitions.csv", index=False)
```
