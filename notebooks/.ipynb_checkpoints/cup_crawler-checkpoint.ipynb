```python
import requests
from bs4 import BeautifulSoup
import csv
import time
import os

input_csv = "../data/competitions.csv"
output_csv = "../data/competition_logos.csv"

scraped_competitions = set()

# Load already scraped competition IDs
if os.path.exists(output_csv):
    with open(output_csv, newline='', encoding='utf-8') as existing_file:
        reader = csv.DictReader(existing_file)
        for row in reader:
            scraped_competitions.add(row['competition_id'])

# Open input and output files
with open(input_csv, newline='', encoding='utf-8') as infile, \
     open(output_csv, mode='a', newline='', encoding='utf-8') as outfile:

    reader = csv.DictReader(infile)
    fieldnames = ['competition_id', 'competition_name', 'logo_url']
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)

    # Write header if the file is new
    if os.stat(output_csv).st_size == 0:
        writer.writeheader()

    for row in reader:
        comp_id = row['competition_id'].strip()
        comp_name = row['name'].strip()
        url = row['url'].strip()

        if not url or comp_id in scraped_competitions:
            print(f"⏭️ Skipping already scraped: {comp_name} ({comp_id})")
            continue

        print(f"🔍 Scraping logo for {comp_name} ({comp_id})...")
        try:
            headers = {"User-Agent": "Mozilla/5.0"}
            response = requests.get(url, headers=headers, timeout=30)

            if response.status_code == 200:
                soup = BeautifulSoup(response.text, 'html.parser')

                # ✅ Try known logo path patterns
                img_tag = soup.find('img', src=lambda src: src and (
                    "/images/erfolge/fix/" in src or
                    "/images/flagge/headerRund/" in src or
                    "/images/logo/header/" in src
                ))

                # ✅ Fallback: match by title/alt text
                if not img_tag:
                    img_tags = soup.find_all('img')
                    for img in img_tags:
                        alt = img.get('alt', '').lower()
                        title = img.get('title', '').lower()
                        if comp_name.lower() in alt or comp_name.lower() in title:
                            img_tag = img
                            break

                if img_tag and 'src' in img_tag.attrs:
                    logo_url = img_tag['src']
                    if logo_url.startswith("//"):
                        logo_url = "https:" + logo_url
                    elif logo_url.startswith("/"):
                        logo_url = "https://www.transfermarkt.com" + logo_url

                    writer.writerow({
                        'competition_id': comp_id,
                        'competition_name': comp_name,
                        'logo_url': logo_url
                    })
                    scraped_competitions.add(comp_id)
                    print(f"✅ Found logo: {logo_url}")
                else:
                    print(f"⚠️ Logo not found for {comp_name}")
            else:
                print(f"❌ Failed to fetch {url} (status {response.status_code})")

        except Exception as e:
            print(f"❌ Error scraping {comp_name}: {e}")

        time.sleep(1)  # Be polite
```
